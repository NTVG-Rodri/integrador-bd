```python
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv("/home/jovyan/work/.env")

conn = mysql.connector.connect(
    host=os.getenv("DB_HOST"),
    port=int(os.getenv("DB_PORT")),
    database=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD")
)
```

```python
with open("/home/jovyan/work/sql/creacion_triggers.sql", "r", encoding="utf-8") as f:
    tr = f.read()
```

```python
""" cursor = conn.cursor()

cursor.execute(tr)

conn.commit()

print("\nSchema ejecutado correctamente") """



cursor = conn.cursor()

for i, statement in enumerate(tr.split("//")):
    stmt = statement.strip()

    if stmt:
        try:
            print(f"\nEjecutando statement {i}")
            cursor.execute(stmt)
        except Exception as e:
            print("\nERROR EN:")
            print(stmt)
            print("\nMENSAJE:")
            print(e)
            break

conn.commit()

print("\nSchema ejecutado correctamente")
```
