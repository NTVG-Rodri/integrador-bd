```python
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv("/home/jovyan/work/.env")

conn = mysql.connector.connect (
    host=os.getenv("DB_HOST"),
    port=int(os.getenv("DB_PORT")),
    database=os.getenv("DB_NAME"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD")
)

print("Conexion exitosa")
```

```python
with open("/home/jovyan/work/sql/schema.sql", "r", encoding="utf-8") as f:
    schema= f.read()
    
print(schema[:500])
```

```python
cursor = conn.cursor()

for i, statement in enumerate(schema.split(";")):
    stmt = statement.strip()

    if stmt:
        try:
            print(f"\nEjecutando statement {i}")
            cursor.execute(stmt)
        except Exception as e:
            print("\nERROR EN:")
            print(stmt)
            print("\nMENSAJE:")
            print(e)
            break

conn.commit()

print("\nSchema ejecutado correctamente")
```
